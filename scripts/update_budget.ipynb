```python
import pandas as pd

# Read the CSV file into a dataframe
csv_path = '/Users/noslen/Documents/Budget/Statements/example_input.csv'
transactions_df = pd.read_csv(csv_path)

# Display the dataframe
print(transactions_df)
```

```python
# Path to the budget.xlsx file
budget_path = '/Users/noslen/Documents/Budget/Budget 2024_v2 copy.xlsx'

# Read the Excel file into a dataframe
budget_df = pd.read_excel(budget_path)

# Display the dataframe
print(budget_df)
```

```python
categories = budget_df.columns[3:18].tolist()
print(categories)
```

```python
import sys

category_mapping = {}

for index, row in transactions_df.iterrows():
    print(f"Transaction {index + 1}: {row['Post Date']} - {row['Description']} - {row['Debit'] if pd.notnull(row['Debit']) else row['Credit']}")
    print("Select a category:")
    for i, category in enumerate(categories, 1):
        print(f"{i}. {category}")
    
    selected_category_index = int(input("Enter the number corresponding to the category: ")) - 1
    selected_category = categories[selected_category_index]
    
    category_mapping[row['Description']] = selected_category

    sys.stdout.flush()

print(category_mapping)
```
